##  Filling Null Values In gender And Avg_Sleep_hour

```python
print(df['gender'].isnull().sum()," Is Missing Values")
print(df['gender'].isnull().mean()*100," % IS Missing ")
```

```python
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
```

```python
print(df['gender'].isnull().sum(),"Is Missing Value")
```

```python
print(df['avg_sleep_hours'].isnull().sum(),"IS Missing Value")
print(df['avg_sleep_hours'].isnull().mean()*100,"%  Missing Value")
```

```python
df['avg_sleep_hours'].hist(bins=30)
df['avg_sleep_hours'].skew()
```

```python
#There Is Some Outliers So We Use Median To Put Missing values

df['avg_sleep_hours'] = df['avg_sleep_hours'].fillna(df['avg_sleep_hours'].median())
```

```python
print(df['avg_sleep_hours'].isnull().sum()," IS Missing VAlue In Avg Sleep Hour")
```
